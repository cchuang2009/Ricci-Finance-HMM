## 1.1 Relation, Forman-Ricci, and Ollivier-Ricci / 三種方法簡要比較

| Method 方法 | Main question 核心問題 | Strength 優點 | Limitation 限制 |
|---|---|---|---|
| Correlation relation network<br> 相關係數網路 | Do two returns move together?<br> 兩檔報酬是否同步？ | Direct, signed, easy to explain <br>直接、具方向、易解釋 | Pairwise only; no local topology <br>僅成對，忽略局部拓樸 |
| Forman-Ricci network | Is an edge supported or overloaded by nearby graph structure?<br> 邊是否受到鄰近結構支撐？ | Very fast; suitable for many rolling frames<br> 計算快，適合大量滾動視窗 | More combinatorial and less transport-aware <br>較偏組合結構，不直接描述質量搬運 |
| Ollivier-Ricci network | How similar are the probability neighborhoods around two endpoints?<br> 兩端點局部分布有多相似？ | Rich local geometry; bridge fragility has intuitive meaning <br>局部幾何豐富，橋接脆弱性直觀 | Slower because optimal transport is solved <br>計算較慢 |

**Network introductions / 網路簡介**

- **Forman-Ricci network:** the original correlation graph is retained, but every edge receives a Forman curvature derived from edge/node weights and adjacent incidence structure. Negative values often indicate relatively overloaded or bridge-like links.<br><b>Forman-Ricci 網路：</b>保留原相關網路，再依節點、邊權重及鄰接結構計算每條邊的 Forman 曲率；負值常表示相對過載或橋接型連結。
- **Ollivier-Ricci network:** each endpoint spreads probability mass over itself and neighbors. The Wasserstein transport distance between the two local measures is compared with their graph distance to obtain curvature.<br><b>Ollivier-Ricci 網路：</b>每個端點把機率質量分配至自身及鄰居，再以 Wasserstein 最優傳輸距離比較兩個局部分布，並相對於圖距離得到曲率。

**Here we use the Olliuvier Ricci, implemented by SciPy. 本節使用解釋力高的 Ollivier-Ricci, scipy-implemented。**

## 2.1. Capital concentration

For normalized capital shares $p_i$, the Herfindahl concentration index is

$$H = \sum_i p_i^2.$$

A larger value means capital is concentrated in fewer nodes or edges.

**本節檢查每條邊的相關、距離、權重與 Ricci 曲率；負曲率邊通常較像橋接或脆弱連結。**


````python
def concentration_index(values) -> float:
    x = np.asarray(list(values), dtype=float)
    x = x[np.isfinite(x)]
    x = np.clip(x, 0.0, None)
    total = float(x.sum())
    if total <= 0:
        return np.nan
    p = x / total
    return round(float(np.sum(p * p)),4)

latest_dv = dollar_volume.iloc[-WINDOW:].mean().dropna()
capital_share = latest_dv / latest_dv.sum()

display(
    pd.DataFrame({
        "mean_dollar_volume": latest_dv,
        "capital_share": capital_share,
    }).sort_values("capital_share", ascending=False)
)
print("Latest-window capital concentration:", concentration_index(latest_dv))
````

````python
progress_log = []

def notebook_progress(done, total, date):
    if done == 1 or done == total or done % max(1, total // 5) == 0:
        message = f"{done:>3}/{total:<3}  {pd.Timestamp(date).date()}"
        progress_log.append(message)
        print(message)

# forman/ollivier_lp"

CURVATURE_ENGINE= "ollivier_lp"
frames, feature_df = build_rolling_frames(
    returns=returns,
    dollar_volume=dollar_volume,
    window=WINDOW,
    step=STEP,
    max_frames=MAX_FRAMES,
    k=K_NEIGHBORS,
    min_corr=MIN_CORR,
    max_bridges=MAX_BRIDGES,
    min_obs=MIN_OBS,
    alpha=OLLIVIER_ALPHA,
    curvature_engine=CURVATURE_ENGINE,
    capital_alpha=CAPITAL_ALPHA,
    use_capital=True,
    progress=notebook_progress,
)

print("Frames:", len(frames))
print("Feature rows:", len(feature_df))
display(feature_df.tail(2).T)
````

````python
def validate_hmm_features(
    frame: pd.DataFrame,
    requested: list[str],
) -> list[str]:
    missing = [name for name in requested if name not in frame.columns]
    if missing:
        raise ValueError(f"Missing HMM features: {missing}")

    usable = []
    report = []

    for name in requested:
        finite_count = np.isfinite(
            pd.to_numeric(frame[name], errors="coerce")
        ).sum()
        report.append({
            "feature": name,
            "finite_rows": int(finite_count),
            "total_rows": len(frame),
        })
        if finite_count > 0:
            usable.append(name)

    display(pd.DataFrame(report))
    return usable

HMM_FEATURES = validate_hmm_features(
    feature_df,
    list(DEFAULT_HMM_FEATURES),
)
print("HMM features:", HMM_FEATURES)
````

Inspect correlation, distance, weight, and Ricci curvature;  本節檢查每條邊的相關、距離、權重與 Ricci 曲率；負曲率邊通常較像橋接或脆弱連結。**


````python
#latest = frames_ollivier[-1]["graph"]
latest_frame = frames[-1]
latest_graph = latest_frame["graph"]
latest_date = pd.Timestamp(latest_frame["date"])

edge_table = pd.DataFrame([
    {
        "source": str(u),
        "target": str(v),
        "correlation": d.get("correlation"),
        "distance": d.get("distance"),
        "weight": d.get("weight"),
        "ricciCurvature": d.get("ricciCurvature"),
        "wassersteinDistance": d.get("wassersteinDistance"),
        "edge_capital_flow": d.get("edge_capital_flow"),
        "engine": d.get(
            "ricci_engine",
            latest_graph.graph.get("curvature_engine"),
        ),
    }
    for u, v, d in latest_graph.edges(data=True)
]).sort_values("ricciCurvature")

print("Latest frame:", latest_date.date())
print("Nodes:", latest_graph.number_of_nodes())
print("Edges:", latest_graph.number_of_edges())
display(edge_table)
````

### 5.1. Node-level capital and curvature table

Make Table about node information, include sector, capital, and ricci curvature. <br>
製作節點的資料表, 包含 產業, 資金, ricci 曲率.     


````python
node_rows = []
for node, attrs in latest_graph.nodes(data=True):
    incident_curvature = [
        float(latest_graph.edges[node, nbr].get("ricciCurvature", np.nan))
        for nbr in latest_graph.neighbors(node)
    ]
    node_rows.append({
        "ticker": str(node),
        "sector": assign_sectors([node])[str(node)],
        "degree": latest_graph.degree(node),
        "capital_share": attrs.get("capital_share"),
        "mean_incident_ricci": np.nanmean(incident_curvature)
            if incident_curvature else np.nan,
    })

node_table = pd.DataFrame(node_rows).sort_values(
    ["capital_share", "degree"],
    ascending=False,
)
display(node_table)
````

## 6. Fit HMM regimes

Use Hidden Markov Method (HMM) to inspect Data Graph, Market state with probability, investment transition, and near future prediction. 

本節以 Gaussian HMM 將滾動圖特徵壓縮為市場狀態，並檢查狀態機率、轉移與預測。**

**Note: ** `GaussianHMM` uses separate `fit()` and `predict()` calls inside the package. It does not call `fit_predict()`.

````python
# features_ollivier/  features_forman

 
if  len(features_ollivier)<max(12,N_STATES*4): 
    features_Ricci = features_forman.copy()
    frames_Ricci = frames_forman.copy()
    print("Use Forman Ricci.")  
else:
    features_Ricci =  features_ollivier.copy()
    frames_Ricci = frames_ollivier.copy() 
    print("use Ollivier Ricci.")


hmm = fit_gaussian_hmm(
    features_Ricci ,
    list(DEFAULT_HMM_FEATURES),
    n_states=N_STATES,
    random_state=42,
)
aligned_frames = [frames_Ricci[i] for i in hmm.valid_index]
aligned_features = features_Ricci.iloc[hmm.valid_index].reset_index(drop=True).copy()
aligned_features["hmm_state"] = hmm.states
regime_names = build_regime_labels(aligned_features, hmm.states)
aligned_features["regime_name"] = [regime_names[int(s)] for s in hmm.states]
aligned_features.tail(2).T
````

````python
gc.collect()

hmm_result = fit_gaussian_hmm(
    feature_df=feature_df,
    features=HMM_FEATURES,
    n_states=N_STATES,
    random_state=RANDOM_STATE,
)

valid_feature_df = feature_df.iloc[hmm_result.valid_index].copy()
valid_feature_df["hmm_state"] = hmm_result.states

regime_names = build_regime_labels(
    valid_feature_df,
    hmm_result.states,
)
valid_feature_df["regime_name"] = [
    regime_names[int(state)]
    for state in hmm_result.states
]

print("Valid HMM rows:", len(valid_feature_df))
print("Regime names:", regime_names)
display(valid_feature_df.tail(2).T)
````

### 6.3 Regime statistics and transition matrix

Statistics and Transition of Regime of Nodes discussed. <br> 
本節以 Gaussian HMM 將滾動圖特徵壓縮為市場狀態，並檢查狀態機率、轉移與預測。

##  7. Perelman-inspired graph surgery

This is a financial-network heuristic, not Perelman's differential-geometric surgery. It removes sufficiently negative-curvature edges and then compares topology before and after removal.

本節將圖手術視為壓力測試，用來觀察移除脆弱邊後的結構變化，而非真實市場切斷預測。


````python
negative_curvatures = edge_table["ricciCurvature"].dropna()
if len(negative_curvatures):
    surgery_threshold = float(negative_curvatures.quantile(0.20))
else:
    surgery_threshold = -1.0

surgery_graph, surgery_info = graph_surgery(
    latest_graph,
    curvature_threshold=surgery_threshold,
    min_component_size=2,
)

surgery_summary = pd.DataFrame([{
    "threshold": surgery_threshold,
    "nodes_before": latest_graph.number_of_nodes(),
    "edges_before": latest_graph.number_of_edges(),
    "components_before": nx.number_connected_components(latest_graph),
    "nodes_after": surgery_graph.number_of_nodes(),
    "edges_after": surgery_graph.number_of_edges(),
    "components_after": nx.number_connected_components(surgery_graph),
    "removed_edges": len(surgery_info.get("removed_edges", [])),
}])

display(surgery_summary)
display(pd.DataFrame(
    surgery_info.get("removed_edges", []),
    columns=["source", "target"],
))
````

## 10. Visualization outputs and conclusions / 視覺化輸出與結論

### Regime feature history / 狀態特徵歷史
- **Output:** rolling average curvature, negative-edge ratio, edge stability, and capital concentration.
- **輸出：**滾動平均曲率、負曲率邊比例、邊穩定度與資金集中度。
- **Conclusion:** simultaneous falling curvature and rising negative-edge ratio indicate increasing structural fragility.
- **結論：**若平均曲率下降且負曲率邊比例上升，代表市場網路脆弱性增強。

### 2D network / 二維網路
- **Output:** node color = sector, node size = capital share, edge width = absolute correlation, edge color = curvature sign.
- **輸出：**節點顏色＝產業、大小＝資金占比、邊寬＝相關絕對值、邊色＝曲率正負。
- **Conclusion:** large nodes joined by thick negative-curvature edges are priority stress links.
- **結論：**大型節點之間若由粗且負曲率的邊連接，應列為優先壓力觀察對象。

### Edge-curvature bar and table / 邊曲率長條圖與表格
- **Output:** correlation, distance, graph weight, curvature, and Ollivier Wasserstein distance.
- **輸出：**相關係數、距離、圖權重、曲率，以及 Ollivier 的 Wasserstein 距離。
- **Conclusion:** inspect the lowest-curvature edges first; they are the least locally redundant.
- **結論：**先檢查曲率最低的邊，因為其局部替代路徑通常最少。

### Sector momentum and flow heatmap / 產業動能與流量熱圖
- **Output:** average sector return momentum and a dimensionless model-derived inter-sector flow score.
- **輸出：**產業平均報酬動能與無量綱的模型產業間流量分數。
- **Conclusion:** positive momentum plus increasing inflow is stronger confirmation than either signal alone.
- **結論：**正動能加上流入增強，比單獨任一訊號更具確認力。

### Capital-flow animation / 資金流動畫
- **Output:** frame-by-frame evolution; color change represents cell-level increase or decrease from the previous frame.
- **輸出：**逐幀演化；顏色變化代表相對前一幀的單格增減。
- **Conclusion:** persistent directional changes matter more than one-frame spikes.
- **結論：**持續性的方向變化比單幀尖峰更重要。

### Graph surgery / 圖手術
- **Output:** before/after stress-test graphs after removing selected negative-curvature edges.
- **輸出：**移除特定負曲率邊前後的壓力測試網路。
- **Conclusion:** this is a sensitivity experiment, not a literal prediction that links vanish.
- **結論：**這是敏感度實驗，不是預測真實市場連結會消失。

### 3D Ricci Galaxy / 三維 Ricci Galaxy
- **Output:** sector direction, capital-share radius, and node-curvature height.
- **輸出：**產業方向、資金占比半徑、節點曲率高度。
- **Conclusion:** sector clusters and curvature outliers reveal structural grouping and exceptions.
- **結論：**產業群聚與曲率離群點可揭示結構分組與異常。

### HMM and GCN / HMM 與 GCN
- **Output:** HMM regime labels, posterior confidence, transition forecast, GCN accuracy, balanced accuracy, and loss.
- **輸出：**HMM 狀態、後驗信心、轉移預測、GCN 準確率、平衡準確率與損失。
- **Conclusion:** GCN performance above the majority baseline suggests graph topology contains useful regime information; it does not independently prove tradable predictability.
- **結論：**若 GCN 高於多數類別基準，表示圖拓樸含有狀態資訊，但不能單獨證明可交易預測能力。

## 12. Final interpretation / 最終解讀

**English conclusion.** Correlation answers *who moves together*. Forman-Ricci quickly screens *which connections are structurally overloaded*. Ollivier-Ricci gives a richer but slower answer about *how similar the two local neighborhoods are*. Use Forman for broad rolling scans and Ollivier for selected frames or suspicious edges. Treat HMM as regime compression, GCN as a topology-information test, and every flow value as a model score rather than a dollar amount.

**中文結論。** 相關係數回答「誰一起漲跌」；Forman-Ricci 快速篩選「哪些連結在結構上較過載」；Ollivier-Ricci 則以較慢但更豐富的方式回答「兩端局部分布有多相似」。大量滾動掃描宜先用 Forman，重點幀或可疑邊再用 Ollivier。HMM 用於壓縮市場狀態，GCN 用於檢驗拓樸是否含狀態資訊；所有資金流數值均應視為模型分數，而非美元金額。
